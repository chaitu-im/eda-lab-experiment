**Discretization**
**and binning**

```python
import pandas as pd
import numpy as np
cricket_jerseys = [18, 7, 45, 10, 12, 33, 99, 77, 88, 63, 5, 24]
jersey_bins = [0, 10, 50, 80, 100]
jersey_categories = pd.cut(cricket_jerseys, jersey_bins)
print("Jersey Categories")
print(jersey_categories)
```

```python
category_counts = pd.Series(jersey_categories).value_counts()
print("Category Counts")
print(category_counts)
```

```python
jersey_bin_names = ['Iconic Single Digits', 'Standard Numbers', 'High Numbers', 'Specialist Numbers']
named_jersey_categories = pd.cut(cricket_jerseys, jersey_bins, labels=jersey_bin_names)
print("Named Jersey Categories")
print(named_jersey_categories)
```

```python
np.random.seed(42)
club_jersey_distribution = pd.cut(np.random.rand(40) * 100, 5, precision=2)
print("Simulated Club Jersey Cut(Bins)")
print(club_jersey_distribution)
```

```python
stadium_jersey_sales = np.random.rand(2000) * 500
sales_quartiles = pd.qcut(stadium_jersey_sales, 4)
print("Stadium Jersey Sales Quartiles")
print(sales_quartiles)
```

```python
qcut_counts = pd.Series(sales_quartiles).value_counts()
print("Sales Count Per Quartile")
print(qcut_counts)
```
