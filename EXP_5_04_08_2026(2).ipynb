**DATA TRANSFORMATION**

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
data = {
    'Matches_Played': [100, 150, 200, 120, 300],
    'Career_Runs': [5000, 6400, 12000, 8500, 120000]
}
df = pd.DataFrame(data)
print("Original Cricket Data:\n", df, "\n")
```

**MIN-MAX SCALING**

```python
min_max_scaler = MinMaxScaler()
df_min_max = pd.DataFrame(
    min_max_scaler.fit_transform(df),
    columns=df.columns
)
print("Min-Max Scaled Cricket Data:\n", df_min_max, "\n")
```

**Z-SCORE STANDARLIZATION**

```python
standard_scaler = StandardScaler()
df_standardized = pd.DataFrame(
    standard_scaler.fit_transform(df),
    columns=df.columns
)
print("Z-Score Standardized Cricket Data:\n", df_standardized, "\n")
```
