```python
import pandas as pd
import numpy as np
df1IPL = pd.DataFrame({'PlayerID': [201, 203, 205, 207, 209, 211, 213, 215, 217, 219, 221],
                        'RunsIPL': [45, 78, 62, 88, 34, 91, 55, 67, 72, 49, 83]})
df2IPL = pd.DataFrame({'PlayerID': [202, 204, 206, 208, 210, 212, 214, 216, 218, 220, 222, 224, 226, 228, 230],
                        'RunsIPL': [61, 74, 39, 82, 95, 47, 58, 63, 71, 90, 36, 79, 52, 66, 84]})

df1WC = pd.DataFrame({'PlayerID': [200, 202, 204, 206, 208, 210, 212, 214, 216, 218, 220, 222, 224, 226, 228],
                       'RunsWC': [58, 64, 71, 39, 82, 90, 47, 65, 73, 58, 91, 44, 67, 80, 52]})
df2WC = pd.DataFrame({'PlayerID': [201, 203, 205, 207, 209, 211, 213, 215, 217, 219],
                       'RunsWC': [77, 63, 85, 49, 91, 38, 66, 72, 59, 81]})
dfIPL = pd.concat([df1IPL, df2IPL], ignore_index=True)
dfWC = pd.concat([df1WC, df2WC], ignore_index=True)
df_opt1 = pd.concat([dfWC, dfIPL], axis=1)
print(df_opt1)
df_opt2 = dfIPL.merge(dfWC, how='inner')
print(df_opt2)
print("Players who featured in BOTH tournaments:", len(df_opt2))
df_opt3 = dfIPL.merge(dfWC, how='left')
print(df_opt3)
df_opt4 = dfIPL.merge(dfWC, how='right')
print(df_opt4)
df_all = dfIPL.merge(dfWC, how='outer')
total_players = df_all['PlayerID'].nunique()
only_ipl = df_all[df_all['RunsWC'].isna()].shape[0]
only_wc = df_all[df_all['RunsIPL'].isna()].shape[0]
both = df_all.dropna(subset=['RunsIPL', 'RunsWC']).shape[0]
print("Total players who featured across both tournaments:", total_players)
print("\n Players who played only in the IPL:", only_ipl)
print("\n Players who played only in the World Cup:", only_wc)
print("\n Players who played in both tournaments:", both)
print("\n")
data = np.arange(15).reshape((3, 5))
indexers = ['Pitch', 'Boundary', 'Crowd']
dframe1 = pd.DataFrame(data, index=indexers,
                        columns=['Wankhede', 'Eden Gardens', 'Chinnaswamy', 'Chepauk', 'Feroz Shah Kotla'])
stacked = dframe1.stack()
stacked.unstack()
print(stacked,"\n")
series1 = pd.Series([4, 6, 1, 0], index=['fours', 'sixes', 'singles', 'dots'])
series2 = pd.Series([100, 50, 200], index=['centuries', 'fifties', 'strikeRateGT150'])
frame2 = pd.concat([series1, series2], keys=['BattingStats', 'MilestoneStats'])
frame2.unstack()
print(frame2,"\n")
```
