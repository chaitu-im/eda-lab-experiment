```python
import numpy as np
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer, SimpleImputer
df = pd.DataFrame(
    {
        "Player": [
            "Virat Kohli",
            "MS Dhoni",
            "Rohit Sharma",
            "KL Rahul",
            "Jasprit Bumrah",
        ],
        "Runs": [630, np.nan, 450, 520, 45],
        "Strike_Rate": [135.5, 142.0, np.nan, 128.4, 95.0],
        "Role": ["Batter", "Wicketkeeper-Batter", "Batter", np.nan, "Bowler"],
    }
).set_index("Player")
mean_imputer = SimpleImputer(strategy="mean")
df["Runs_Mean"] = mean_imputer.fit_transform(df[["Runs"]])
median_imputer = SimpleImputer(strategy="median")
df["Strike_Rate_Median"] = median_imputer.fit_transform(df[["Strike_Rate"]])
mode_imputer = SimpleImputer(strategy="most_frequent")
df["Role_Mode"] = mode_imputer.fit_transform(df[["Role"]]).flatten()
num_cols = ["Runs", "Strike_Rate"]
mice_imputer = IterativeImputer(max_iter=10, random_state=42)
df_mice = df[num_cols].copy()
df_mice.iloc[:, :] = mice_imputer.fit_transform(df_mice)
df["Runs_MICE"] = df_mice["Runs"]
df["Strike_Rate_MICE"] = df_mice["Strike_Rate"]
print(df)
```
